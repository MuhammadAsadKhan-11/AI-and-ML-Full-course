convert emotions into numbers


```python
unique_emotions = df['emotion'].unique()
```

```python
emotion_num={}
i=0
for emotion in unique_emotions:
    emotion_num[emotion]=i
    i+=1
```

```python
df['emotion_num'] = df['emotion'].map(emotion_num)
```

Removing punctuations

```python
import string
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
df['text'] = df['text'].apply(remove_punctuation)
```

```python
# Removing numbers from the text
def remove_numbers(text):
    new=""
    for i in text:
        if not i.isdigit():
            new=new+i
    return new
df['text'] = df['text'].apply(remove_numbers)
```

```python
# REMOVE imojis
def remove_imojis(text):
    new=""
    for i in text:
        if i.isascii():
            new=new+i
    return new
df['text'] = df['text'].apply(remove_imojis)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
# now check the text column and remove the stop words from the text column
def remove(txt):
    word_tokens = word_tokenize(txt)
    cleaned=[]
    for i in word_tokens:
        if i not in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)

```

```python
df['text'] = df['text'].apply(remove)
```

```python
# bag of words model
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

documents=["I love pizza","pizza is good", "I love pasta", "pasta is good"]

x=vectorizer.fit_transform(documents)
print("Vocabulary:\n",vectorizer.get_feature_names_out())
print("Bag of words representation:\n",x.toarray())
```

```python
y=vectorizer.fit_transform(df['text'])
print("Vocabulary for the dataset:\n",vectorizer.get_feature_names_out())
print("Bag of words representation for the dataset:\n",y.toarray())
```
